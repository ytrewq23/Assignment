```python
import numpy as np
np.random.seed(42)

n = 1000000

data = np.column_stack([
    np.random.randint(1, 100001, n),
    np.random.randint(1, 10001, n),
    np.random.randint(500, 100000, n),
    np.random.randint(1, 10, n),
    np.random.randint(0, 31, n),
    np.random.randint(1, 5, n)
])

print("Өлшемі:", data.shape)
print("\nБастапқы:",data[:10])
print("\nСоңғы:",data[-10:])
print(data[100000:100010])
```

```python
total_income = np.sum(data[:, 2])

print("Жалпы табыс:", total_income)
```

```python
region_sales = []

for region in range(1, 5):
    total = np.sum(data[data[:, 5] == region, 2])
    region_sales.append(total)
    print(f"{region}:", total)
```

```python
unique_client_ids, inverse_indices, counts = np.unique(data[:, 0], return_inverse=True, return_counts=True)
sum_purchases_per_client = np.bincount(inverse_indices, weights=data[:, 2])
client_average = sum_purchases_per_client / counts

overall_client = np.mean(client_average)

print("Клиенттердің орташа сатып алу сомасы:", overall_client)
```

```python
clients = data[:, 0]
purchases = data[:, 2]

unique_clients, inverse = np.unique(clients, return_inverse=True)

client_sum = np.bincount(inverse, weights=purchases)
client_count = np.bincount(inverse)

client_average = client_sum / client_count

print("10 клиенттің орташа сатып алу сомасы:")
print(client_average[:10])
```

```python
overall_client_average = np.mean(client_average)

print("Орташа сомасы:", overall_client_average)
```
